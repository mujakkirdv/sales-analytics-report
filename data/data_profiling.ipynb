```python
"""
YData Profiling Report Generator
=================================
Generates a comprehensive HTML profiling report from the `sales` table
in PostgreSQL.

Run:  python profiling_report.py
Output: output.html
"""

import os
import pandas as pd
from dotenv import load_dotenv
from sqlalchemy import create_engine, inspect
from ydata_profiling import ProfileReport

# ---------------------------------------------------------------------------
# LOAD ENVIRONMENT VARIABLES
# ---------------------------------------------------------------------------
load_dotenv()

DB_USER     = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_HOST     = os.getenv("DB_HOST")
DB_PORT     = os.getenv("DB_PORT")
DB_NAME     = os.getenv("DB_NAME")
DB_TABLE    = os.getenv("DB_TABLE", "sales")

# ---------------------------------------------------------------------------
# CONNECT & LOAD DATA
# ---------------------------------------------------------------------------
print("🔌 Connecting to PostgreSQL...")
engine = create_engine(
    f"postgresql+psycopg2://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

# Verify table exists before trying to load it
inspector = inspect(engine)
tables = inspector.get_table_names()
print(f"📋 Tables found: {tables}")

if DB_TABLE not in tables:
    raise ValueError(
        f"❌ Table '{DB_TABLE}' not found in database '{DB_NAME}'. "
        f"Available tables: {tables}"
    )

print(f"📥 Loading table '{DB_TABLE}'...")
df = pd.read_sql_table(DB_TABLE, engine, schema="public")
print(f"✅ Loaded {len(df):,} rows × {df.shape[1]} columns")

# ---------------------------------------------------------------------------
# GENERATE PROFILE REPORT (only once!)
# ---------------------------------------------------------------------------
print("🔍 Generating profiling report...")
profile = ProfileReport(
    df,
    title="Sales Dataset — Profiling Report",
    explorative=True,
    minimal=False,
    config_file=None
)

# ---------------------------------------------------------------------------
# SAVE HTML
# ---------------------------------------------------------------------------
OUTPUT_FILE = "output.html"
profile.to_file(OUTPUT_FILE)
print(f"✅ Report generated: {OUTPUT_FILE}")

# ---------------------------------------------------------------------------
# OPTIONAL: Open in browser (Windows/Mac/Linux)
# ---------------------------------------------------------------------------
try:
    import webbrowser
    webbrowser.open(f"file://{os.path.abspath(OUTPUT_FILE)}")
except Exception:
    pass
```
